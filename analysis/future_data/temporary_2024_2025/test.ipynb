```python
import pandas as pd


for model in ["moirai", "chronos", "time_moe"]:
    # Load raw CSV
    file_path = f"analysis/future_data/temporary_2024_2025/es_future_final_{model}_2024_2025_new.csv"
    df = pd.read_csv(file_path)
    # Identify the datetime column
    datetime_col = [col for col in df.columns if 'datetime' in col.lower()]
    if not datetime_col:
        raise ValueError("No datetime-like column found")
    datetime_col = datetime_col[0]

    # Parse datetime and force UTC interpretation
    df[datetime_col] = pd.to_datetime(df[datetime_col], utc=True)

    # Convert to America/Chicago
    df[datetime_col] = df[datetime_col].dt.tz_convert('America/Chicago')

    df.to_csv(file_path, index=False)
```
